## Gradient Descent:
   ## It is a method used to (Reduces Error ) by moving (Step by Step)
   ## Finding the Best Solution by correcting mistakes little by little.
   
## Types of Gradient Descent
   ## Batch Gradient Descent
   ## Stochastic Gradient Descent
   ## Mini - Batch Gradient Descent


   ## Batch Gradient Descent: It use the entire dataset & then for one time calculate loss by taking all predictions and all actual value difference.

   ## Stochastic Gradient Descent: In Stochastic Gradient Descent with (Every-Row) of Data , we calculate loss and we update weights

   ## Mini-Batch Gradient Descent: It is the combination of Batch and Stochastic Gradient Descent
