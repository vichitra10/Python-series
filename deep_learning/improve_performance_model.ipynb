## Regularization and Batch Normalization
## Regularization is a set of techniques used to reduce overfitting.
## The idea is to prevent the model from becoming unnecessarily complex and memorizing the training data.

    ## Common types
    ## L1 Regularization: Adds a penalty based on the absolute value of weights:

    ## $$ Loss_{new} = Loss + \lambda \sum |w| $$

    ## It can encourage some weights to become zero, which can help with feature selection.

    ## L2 Regularization

    ## Adds a penalty based on squared weights:

    ## $$ Loss_{new} = Loss + \lambda \sum w^2 $$

    ## It encourages the model to keep weights smaller.

## Dropout is another commonly used regularization technique.
