## Vanishing Gradient Problem in Neural Network (Imp for Interview Purpose)

## Vanishing - Gradient Problem: It happens when (Gredients) become (so small ) that earliar layer (Stop-Learning) 
## The network forgots how to learn in the begning layer


## Why does Vanishing Gradient happens?
## Repeated multiplication of small gradients during backpropagation makes the gradient approach zero.
## Main problem:
## Earlier layers receive very small updates → learning becomes very slow.
## Common activation associated with it:
## Sigmoid and tanh, especially when their derivatives are small.
## How it is addressed:
## ReLU activation
## Better weight initialization
## Batch Normalization
## Residual connections (ResNet)
