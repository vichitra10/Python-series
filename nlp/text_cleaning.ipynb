## Removing Emoji: spam detection / Topic Modeling
u"\U0001F300-\U0001F5FF"  ## Emoticons


```python
from textblob import TextBlob
from nltk.corpus import stopwords
import re


## Removing Emoji from the text
def remove_emojis(text):
    emoji_pattern = re.compile(
        "["
        u"\U0001F600-\U0001F64F"  # Emoticons
        u"\U0001F300-\U0001F5FF"  # Symbols & pictographs
        u"\U0001F680-\U0001F6FF"  # Transport & map symbols
        u"\U0001F1E0-\U0001F1FF"  # Flags
        "]+",
        flags=re.UNICODE
    )

    return emoji_pattern.sub(r'', text)


emoji_data = {
'content': [
    "I love pizza 🍕 so much!",
    "Good morning 🌞 🌻",
    "Let's go for a ride 🚗 🏍️",
    "Vacation time 🏖️ ✈️",
    "Happy birthday 🎂 🎉",
    "Feeling strong 💪 🔥",
    "I love to eat 🍔 🍟",
    "This movie is amazing 😍 🎬",
    "I am feeling happy 😊 ❤️",
    "Let's party tonight 🥳 🎉",
    "The weather is beautiful ☀️ 🌈",
    "I am very tired 😴 💤",
    "That's so funny 😂 🤣",

]}
    
df = pd.DataFrame(emoji_data)
df['content'] = df['content'].apply(remove_emojis)
df['content']


```
