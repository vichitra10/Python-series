```python
import tensorflow as tf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OneHotEncoder
from google.colab import files


## 1)  First Step:  Data Gathering
data = pd.read_csv('/content/sample_data/huge_1M_titanic.csv')

## 2)  Second Step: Data Preprocessing

# data.head()
# data.shape ## get total no of rows in the dataset
# data.info() ## get the column datatype
# data.isnull().sum()
dataa = data.drop(columns=['PassengerId','Name','Age','Ticket','Cabin'])
dataa.isnull().sum()

## 3) Third Step) Data Cleaning

dataa.dropna(inplace=True)
# Reset the index after dropping rows to ensure proper alignment during concatenation
dataa.reset_index(drop=True, inplace=True)
dataa.isnull().sum()

dataa['Embarked'].mode()
dataa['Embarked'] = dataa['Embarked'].replace({'S':'Southampton','C':'Chebourg','Q':'QueensTown'})
dataa.head()

## 4) Feature Encoding
## Encode - Gender --> LabelEncoder

label = LabelEncoder()
dataa['Sex'] = label.fit_transform(dataa['Sex'])
dataa.head()

onehot = OneHotEncoder(sparse_output=False)
onehot_encoded_data = onehot.fit_transform(dataa[['Embarked']])
onehot_encoded_data
# dataa.head()
embarked = pd.DataFrame(onehot_encoded_data,columns=[onehot.get_feature_names_out()])
embarked.head()
dataa = pd.concat([dataa.drop(columns=['Embarked']), embarked], axis=1)
# dataa = pd.concat([dataa,embarked])
dataa['Fare'] = dataa['Fare'].astype(int)
dataa.head()

##5)  Feature Scaling
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
num_columns = ['Pclass','SibSp','Parch','Fare']
dataa[num_columns] = scaler.fit_transform(dataa[num_columns])
dataa.head()

## 6)  Train Test split
import pickle
with open('label.pkl','wb') as file:
  pickle.dump(label,file)

with open('onehot_encoded_data','wb') as file:
  pickle.dump(onehot,file)

with open('scaler.pkl', 'wb') as file:
    pickle.dump(scaler, file)


X = dataa.drop(columns='Survived')
Y = dataa['Survived']


## Train-Validation-Test Split
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=20,random_state=42)
X_train,X_val,Y_train,Y_val = train_test_split(X_train,Y_train, test_size=0.2,random_state = 42)
# X_train.shape,X_val.shape,X_test.shape
# Y_train.shape,Y_val.shape,Y_test.shape




```

```python
## Model Building Approach



import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
from tensorflow.keras.callbacks import EarlyStopping


tf.keras.backend.clear_session()

(X_train.shape[1])
model = Sequential([
    Input(shape=(X_train.shape[1],)),
    Dense(64, activation='relu'),
    Dense(32, activation='relu'),
    Dense(1, activation='sigmoid')
])
model.summary()

## Model Compilation ---> Optimizer , Loss, Metrics
opt = tensorflow.keras.optimizers.Adam(learning_rate= 0.01)
loss = tf.keras.losses.BinaryFocalCrossentropy()
model.compile(optimizer='adam',loss='binary_crossentrophy',metrics=['accuracy'])
stopping_callback = EarlyStopping(monitor = 'val_loss', patience=5,verbose=1,restore_best_weights=2)
model.fit(X_train,X_val,Y_train,Y_val,epochs=10, callbacks=stopping_callback)




```
