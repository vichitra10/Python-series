```python
from transformers import AutoModelForCausalLM
import torch
model = AutoModelForCausalLM.from_pretrained(
    'google/gemma-4-31B-it',
    torch_dtype=torch.bfloat16
)
input_token = tokenizer('Write a python code for adding two numbers', return_tensors='pt')
output = model(input_ids = input_token['input_ids'])
gen_out = model.generate(input_ids=input_token['input_ids'], max_new_tokens=100)

tokenizer.decode(gen_out)

```
